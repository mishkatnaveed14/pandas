```python
import pandas as pd
```

```python
# Task # 1
data = {
    "Name" : ["Ali","Sara","Ahmad","Ayesha","Zain"],
     "Marks" : [70,85,60,95,45]
    }
print("Student Data:")
df = pd.DataFrame(data)
print(df)
print("\nAverage Marks:")
print(df["Marks"].mean())
print("\nMaximum Marks:")
print(df["Marks"].max())
print("\nMinimum Marks:")
print(df["Marks"].min())
print("\n Total students:")
print(df["Name"].count())
```

```python
# task # 2
greater_marks = df[df["Marks"] >= 50]
print("Students with marks greater than or equal to 50:")
print(greater_marks)
less_marks = df[df["Marks"] <= 50]
print("\nStudents with marks less than or equal to 50: ")
print(less_marks)
```

```python
# task # 3
df["City"] = ["Karachi", "Lahore", "Karachi", "Islamabad","Lahore"]
df
karachi_students = df[df["City"] == "Karachi"]
print("Karachi Students:")
print(karachi_students)
df
```

```python
# task # 4
data = {
    "Name" : ["Ali","Sara","Ahmad","Ayesha","Zain"],
     "Marks" : [70,85,60,95,45],
"Course" : ["AI","Web","AI","AI","Web"]
    }
df = pd.DataFrame(data)
average_marks = df.groupby("Course")["Marks"].mean()
print("Average Marks by Course:")
print(average_marks)
print("\nStudent Data:")
df
```

```python
# task # 5
data = {
  "Name" : ["Ali","Sara","Ahmad","Ayesha","Zain"],
  "Marks" : [80,None,70,None,90]
}
df = pd.DataFrame(data)
null_values = df.isnull().sum()
print("Null Values:")
print(null_values)
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df
```
